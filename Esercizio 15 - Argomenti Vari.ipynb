```python
#Obiettivo: Riconoscere e convertire tipi di dato diversi.
#• Crea una lista contenente almeno 8 valori: interi, float, stringhe, booleani.
#• Scorri la lista e stampa per ogni elemento: valore, tipo tramite type().
#• Conta quanti elementi appartengono a ciascun tipo.
#• Gestisci separatamente bool e int: in Python bool è una sottoclasse di int, quindi fai attenzione all'ordine dei controlli.

lista = [10, 3.14, "Ciao", True, 25, 7.5, "Python", False]

cont_int = 0
cont_float = 0
cont_str = 0
cont_bool = 0

for elemento in lista:
    if type(elemento) == int:
        cont_int += 1
    elif type(elemento) == float:
        cont_float += 1
    elif type(elemento) == str:
        cont_str += 1
    elif type(elemento) == bool:
        cont_bool += 1

    print(elemento, type(elemento))
    
print(cont_int, cont_float, cont_str, cont_bool)

```
