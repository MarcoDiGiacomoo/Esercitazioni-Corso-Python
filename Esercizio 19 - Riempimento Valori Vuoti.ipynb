```python
#ESERCIZIO 1 - RIEMPIMENTO VALORI VUOTI

import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

dati = {
    "Età": [
        25, 31, None, 45, 29,
        None, 52, 38, 41, None
    ],
    "Peso": [
        68.5, None, 82.3, 77.0, None,
        90.2, 85.1, None, 73.4, 69.8
    ],
    "Altezza": [
        175, 168, 182, None, 170,
        185, None, 178, 172, None
    ],
    "Pressione": [
        120, None, 135, 128, 118,
        None, 140, 125, None, 122
    ],
    "Colesterolo": [
        190, 210, None, 225, 180,
        205, None, 195, 215, None
    ]
}

df = pd.DataFrame(dati)

df.head()
```

```python
df.isna().sum()
```

```python
imputer = KNNImputer(n_neighbors=2)
imputed = pd.DataFrame(imputer.fit_transform(df), columns = df.columns)
imputed
```

```python
# ESERCIZIO 2
# Crea un DataFrame a partire dal seguente dizionario. Individua i valori mancanti e sostituiscili con il valore 0. Infine verifica che non siano rimasti valori nulli.

dati_es1 = {
    "Vendite": [120, None, 95, 140, None, 180, 110, None, 160, 130]
}

df1 = pd.DataFrame(dati_es1)
df1
```

```python
df1["Vendite"] = df1["Vendite"].fillna(0)
df1
```

```python
# Crea un DataFrame a partire dal seguente dizionario. Calcola:

# la media della colonna Eta e usala per sostituire i valori mancanti;
# la mediana della colonna Stipendio e usala per sostituire i valori mancanti;
# la moda della colonna Citta e usala per sostituire i valori mancanti.

# Verifica infine che il DataFrame non contenga più valori nulli.

dati_es2 = {
    "Eta": [25, 30, None, 45, 32, None, 50, 28, 35, None],
    "Stipendio": [1500, 1800, 1700, None, 2000, 1750, None, 1900, 1850, 1780],
    "Citta": ["Roma", "New York", "New York", None, "Parigi", "New York", "Roma", None, "New York", "New York"]
}

df2 = pd.DataFrame(dati_es2)
df2
```

```python
df2.isna().sum()
```

```python
df2["Eta"] = df2["Eta"].fillna(df2["Eta"].mean())
df2["Stipendio"] = df2["Stipendio"].fillna(df2["Stipendio"].median())
df2["Citta"] = df2["Citta"].fillna(df2["Citta"].mode()[0])

df2
```

```python
# Crea un DataFrame a partire dal seguente dizionario. 
# Sostituisci i valori mancanti della colonna Peso con la media della colonna e quelli della colonna Colesterolo con la mediana. Lascia invariata la colonna Nome.

dati_es3 = {
    "Nome": [
        "Mario", "Luca", "Giulia", "Anna", "Paolo",
        "Sara", "Marco", "Elena", "Davide", "Chiara"
    ],
    "Peso": [
        68.5, None, 82.3, 77.0, None,
        90.2, 85.1, None, 73.4, 69.8
    ],
    "Colesterolo": [
        190, 210, None, 225, 180,
        205, None, 195, 215, None
    ]
}

df3 = pd.DataFrame(dati_es3)
df3
```

```python
df3["Peso"] = df3["Peso"].fillna(df3["Peso"].mean())
df3["Colesterolo"] = df3["Colesterolo"].fillna(df3["Colesterolo"].median())
df3
```

```python
# Crea un DataFrame a partire dal seguente dizionario e utilizza KNNImputer di scikit-learn per stimare i valori mancanti.
# Prova almeno due valori diversi di n_neighbors (ad esempio 3 e 5) e confronta i risultati ottenuti.

dati_es4 = {
    "Peso": [68, 72, None, 80, 77, None, 65, 90, 85, 70],
    "Altezza": [175, 180, 178, None, 182, 170, None, 188, 185, 176],
    "Eta": [25, 31, 29, 40, None, 35, 22, 45, None, 28]
}

df4 = pd.DataFrame(dati_es4)
df4
```
