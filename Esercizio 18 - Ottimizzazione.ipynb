```python
import numpy as np
import pandas as pd
```

```python
#crea un dataframe con una colonna di temperature in gradi Celsius e aggiungi una colonna con la loro conversione in fahreneit (°C × 1,8) + 32

dati = {"Gradi_Celsius": [18, 25, 21, 31, 27]}

df = pd.DataFrame(dati)

df["Gradi_Fahreneit"] = df["Gradi_Celsius"] * 1.8 + 32
df
```

```python
#hai una colonna con punteggi numerici. crea una nuova colonna che indichi promosso se il punteggio >= 18 oppure bocciato

dati = {"Punteggio": [18, 15, 27, 30, 11, 17, 23]}

df = pd.DataFrame(dati)
df["Esito"] = np.where(df["Punteggio"] >= 18, "Promosso", "Bocciato")
df
```

```python
#hai due colonne orario_di_lavoro e paga_oraria. crea una nuova colonna che calcoli la paga settimanale, e applichi un bonus del 10% se le ore lavorate superano le 40

dati = {"Orario_di_Lavoro": [18, 40, 40, 25, 42, 50], "Paga_Oraria": [9, 9, 9, 9, 9, 9]}

df = pd.DataFrame(dati)
df["Paga_Settimanale"] = np.where(df["Orario_di_Lavoro"] > 40, df["Orario_di_Lavoro"] * df["Paga_Oraria"] * 1.1, df["Orario_di_Lavoro"] * df["Paga_Oraria"])
df
```
