## Parte 3 – Elaborazione dei dati

### Creazione di nuove colonne

7. Creare una colonna `Incasso` calcolata come:

   `Quantita * Prezzo`

8. Creare una colonna `Ordine_Grande`:

   - `True` se la quantità acquistata è maggiore o uguale a 3.
   - `False` altrimenti.

9. Creare una colonna `Fascia_Prezzo`:

   - Prezzo inferiore a 100 € → `Economico`
   - Prezzo da 100 € a 500 € → `Medio`
   - Prezzo superiore a 500 € → `Premium`


```python
df["Incasso"] = df["Quantita"] * df["Prezzo"]
df.head()
```

```python
df["Ordine_Grande"] = df["Quantita"] >= 3
df.head()
```

```python
def fascia_prezzo(prezzo):
    if prezzo < 0:
        raise ValueError("Errore: Prezzo Negativo") 
    elif prezzo < 100:
        return "Economico"
    elif prezzo >= 100 and prezzo < 500:
        return "Medio"
    else:
        return "Premium"

df["Fascia_Prezzo"] = df["Prezzo"].apply(fascia_prezzo)
df.head()
```

## Parte 4 – Analisi generale delle vendite

Calcolare e visualizzare:

10. Il fatturato totale.
11. Il numero totale di unità vendute.
12. Il prezzo medio dei prodotti venduti.
13. Il valore medio degli ordini.
14. Il prodotto più costoso.
15. Il prodotto meno costoso.
16. Il numero di ordini con quantità maggiore o uguale a 3.
17. La percentuale di ordini grandi rispetto al totale.

```python
#Il fatturato totale.
#Il numero totale di unità vendute.
#Il prezzo medio dei prodotti venduti.
#Il valore medio degli ordini.

fatturato_tot = df["Incasso"].sum()
unita_vendute = df["Quantita"].sum()
prezzo_medio = df["Prezzo"].mean()
valore_medio_ordini = df["Incasso"].mean()

print(f"Fatturato Totale {fatturato_tot.round(2)}")
print(f"Numero Totale di Unità Vendute: {unita_vendute.round(2)}")
print(f"Prezzo Medio Prodotti Venduti: {prezzo_medio.round(2)}")
print(f"Valore Medio Ordini: {valore_medio_ordini.round(2)}")
```

```python
#Il prodotto più costoso.
#Il prodotto meno costoso.

print("PRODOTTO PIU' COSTOSO")
print(df.loc[df["Prezzo"].idxmax(), ["Prodotto", "Prezzo"]])
print()
print("PRODOTTO MENO COSTOSO")
print(df.loc[df["Prezzo"].idxmin(), ["Prodotto", "Prezzo"]])
```

```python
#Il numero di ordini con quantità maggiore o uguale a 3.
#La percentuale di ordini grandi rispetto al totale.

numero_ordini_grandi = df[df["Ordine_Grande"] == True]["Ordine_ID"].value_counts().sum()
print(f"Numero Ordini Grandi: {numero_ordini_grandi}")

totale_ordini = df["Ordine_ID"].value_counts().sum()

perc_ordini_grandi = round(numero_ordini_grandi / totale_ordini * 100, 2)

print(f"Percentuale Ordini grandi sul Totale: {perc_ordini_grandi}")
```

```python
#Parte 5 – Analisi per categoria e prodotto
#Calcolare il fatturato totale per ogni categoria.

fatturato_tot_cat = df.groupby("Categoria")["Incasso"].sum()

#Calcolare la quantità totale venduta per ogni prodotto.

quantita_tot_prod = df.groupby("Prodotto")["Quantita"].sum()

#Calcolare il fatturato totale per ogni prodotto.

fatturato_tot_prod = df.groupby("Prodotto")["Incasso"].sum()

#Ordinare i prodotti in base al fatturato, dal più alto al più basso.

prodotti_per_fattura = df.groupby("Prodotto")["Incasso"].sum().sort_values(ascending=False)

#Individuare la categoria con il fatturato maggiore.

cat_fatturato_maggiore = df.groupby("Categoria")["Incasso"].sum()

#Individuare il prodotto più venduto in termini di quantità.
#Mostrare solamente i prodotti che hanno generato un fatturato superiore a 1.000 €.

print(f"Fatturato totale per ogni categoria: {fatturato_tot_cat}")
print()
print(f"Quantità totale venduta per ogni prodotto: {quantita_tot_prod}")
print()
print(f"Fatturato totale per ogni prodotto: {fatturato_tot_prod}")
print()
print(f"Prodotti ordinati in base al fatturato, dal più alto al più basso: {prodotti_per_fattura}")
print()
print(f"Categoria con il fatturato maggiore: {cat_fatturato_maggiore.idxmax()} {cat_fatturato_maggiore.max()}")


df.head()
```

```python
# Parte 6 – Analisi dei metodi di pagamento
#Calcolare il numero di ordini effettuati con ciascun metodo di pagamento.

conteggio_metodo_pag = df["Pagamento"].value_counts()

#Calcolare il fatturato generato da ciascun metodo di pagamento.

fatturato_metodo_pag = df.groupby("Pagamento")["Incasso"].sum()

#Individuare il metodo di pagamento che ha generato il fatturato maggiore.

metodo_pag_fatturato_maggiore = df.groupby("Pagamento")["Incasso"].sum()

#Calcolare la percentuale di ordini pagati tramite Carta rispetto al totale.
ordini_carta = df[df["Pagamento"] == "Carta"].value_counts().sum()
perc_ordini_carta = round(ordini_carta / df["Pagamento"].value_counts().sum() * 100,2)


print(f"Numero di ordini effettuati con ciascun metodo di pagamento: {conteggio_metodo_pag}")
print()
print(f"Fatturato generato da ciascun metodo di pagamento: {fatturato_metodo_pag}")
print()
print(f"Metodo di pagamento che ha generato il fatturato maggiore: {metodo_pag_fatturato_maggiore.idxmax()} {metodo_pag_fatturato_maggiore.max()}")
print()
print(f"Percentuale di ordini pagati tramite Carta rispetto al totale: {perc_ordini_carta}%")
```
